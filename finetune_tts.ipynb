```python
import os
import math
import torch
import numpy as np
from dataclasses import dataclass
from typing import List, Dict, Any

from datasets import load_dataset, Dataset
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    TrainingArguments,
    Trainer,
)
from peft import LoraConfig, get_peft_model, TaskType
```

```python
MODEL_NAME = "SPRINGLab/Indic-Mio"
CODEC_NAME = "Aratako/MioCodec-25Hz-44.1kHz-v2"   # matches Indic-Mio's own codec

SPEECH_OFFSET = 151669   # first speech-token id in the LLM vocab
NUM_CODES = 12800        # codec codebook size

MAX_SEQ_LEN = 1024        # keep in line with the --max-model-len you served with
OUTPUT_DIR = "/content/indic-mio-lora"
DATASET_NAME = "SPRINGLab/IndicTTS-Hindi"
TEXT_COLUMN = "text"
AUDIO_COLUMN = "audio"

device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
from miocodec import MioCodecModel, load_audio

print("MioCodec imported successfully!")
print(MioCodecModel)
```

```python
print("Loading MioCodec...")
codec = MioCodecModel.from_pretrained(CODEC_NAME).eval().to(device)
codec_sample_rate = codec.config.sample_rate


def encode_audio_to_codes(audio_array: np.ndarray, sampling_rate: int) -> List[int]:
    """Encode one waveform to a flat list of codec content-token ids."""
    import soundfile as sf
    import tempfile

    with tempfile.NamedTemporaryFile(suffix=".wav") as tmp:
        sf.write(tmp.name, audio_array, sampling_rate)
        waveform = load_audio(tmp.name, sample_rate=codec_sample_rate).to(device)

    with torch.inference_mode():
        features = codec.encode(waveform)

    return features.content_token_indices.squeeze().tolist()
```

```python
print("Loading tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, trust_remote_code=True)
if tokenizer.pad_token_id is None:
    tokenizer.pad_token = tokenizer.eos_token

print(f"Loading dataset {DATASET_NAME}...")
raw_dataset = load_dataset(DATASET_NAME, split="train")


def build_example(example: Dict[str, Any]) -> Dict[str, Any]:
    text = example[TEXT_COLUMN]
    audio = example[AUDIO_COLUMN]

    messages = [{"role": "user", "content": text}]
    prompt = tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    prompt_ids = tokenizer(prompt, add_special_tokens=False)["input_ids"]

    codec_codes = encode_audio_to_codes(audio["array"], audio["sampling_rate"])
    speech_ids = [SPEECH_OFFSET + c for c in codec_codes if 0 <= c < NUM_CODES]

    input_ids = prompt_ids + speech_ids + [tokenizer.eos_token_id]
    labels = [-100] * len(prompt_ids) + speech_ids + [tokenizer.eos_token_id]

    return {"input_ids": input_ids, "labels": labels}


print("Encoding dataset (this calls the codec on every clip, so it can take a while)...")
processed = raw_dataset.map(
    build_example,
    remove_columns=raw_dataset.column_names,
    desc="encoding audio -> codec tokens",
)

processed = processed.filter(lambda ex: len(ex["input_ids"]) <= MAX_SEQ_LEN)
print(f"Kept {len(processed)} / {len(raw_dataset)} examples after length filtering.")
```

```python
@dataclass
class SpeechSFTCollator:
    pad_token_id: int

    def __call__(self, batch: List[Dict[str, Any]]) -> Dict[str, torch.Tensor]:
        max_len = max(len(b["input_ids"]) for b in batch)

        input_ids, labels, attn_mask = [], [], []
        for b in batch:
            pad_len = max_len - len(b["input_ids"])
            input_ids.append(b["input_ids"] + [self.pad_token_id] * pad_len)
            labels.append(b["labels"] + [-100] * pad_len)
            attn_mask.append([1] * len(b["input_ids"]) + [0] * pad_len)

        return {
            "input_ids": torch.tensor(input_ids, dtype=torch.long),
            "attention_mask": torch.tensor(attn_mask, dtype=torch.long),
            "labels": torch.tensor(labels, dtype=torch.long),
        }


collator = SpeechSFTCollator(pad_token_id=tokenizer.pad_token_id)
```

```python
print("Loading base model...")
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.bfloat16,
    device_map="auto",
    trust_remote_code=True,
)
model.config.use_cache = False  # required with gradient checkpointing
model.gradient_checkpointing_enable()

lora_config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    r=16,
    lora_alpha=32,
    lora_dropout=0.05,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    bias="none",
)
model = get_peft_model(model, lora_config)
model.print_trainable_parameters()
```

```python
training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,

    # T4-friendly settings
    per_device_train_batch_size=1,
    gradient_accumulation_steps=16,

    num_train_epochs=3,
    learning_rate=1e-4,

    lr_scheduler_type="cosine",

    # Replace warmup_ratio
    warmup_steps=100,

    logging_steps=10,

    save_strategy="epoch",
    save_total_limit=2,

    fp16=True,
    report_to="none",
)
```

```python
adapter_dir = f"{OUTPUT_DIR}/adapter"
model.save_pretrained(adapter_dir)
tokenizer.save_pretrained(adapter_dir)
print("LoRA adapter saved to", adapter_dir)

# Merge for a standalone checkpoint you can `vllm serve` directly:
merged_dir = f"{OUTPUT_DIR}/merged"
merged_model = model.merge_and_unload()
merged_model.save_pretrained(merged_dir)
tokenizer.save_pretrained(merged_dir)
print("Merged model saved to", merged_dir)
```
